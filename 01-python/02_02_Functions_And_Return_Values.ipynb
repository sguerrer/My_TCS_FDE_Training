## 1. Define and call

```python
def greet(name):
    print(f"Hello, {name}!")


greet("Ada")
greet("Grace")
```

## 2. Parameters

A parameter is the name in the definition. An argument is the value you pass in.

```python
def describe_model(name, max_tokens=1024, temperature=1.0):
    print(f"{name}: max_tokens={max_tokens}, temperature={temperature}")


describe_model("sonnet")                         # use defaults
describe_model("sonnet", 2048)                   # positional
describe_model("sonnet", temperature=0.2)        # keyword
describe_model(temperature=0, name="haiku")      # keywords in any order
```

### `*args` and `**kwargs`

```python
def total(*numbers):
    return sum(numbers)


def show_options(**options):
    for key, value in options.items():
        print(f"{key} = {value}")


print(total(1, 2, 3, 4))
show_options(model="sonnet", stream=True)
```

```python
# The * and ** operators also unpack values at call time
numbers = [10, 20, 30]
settings = {"model": "haiku", "stream": False}

print(total(*numbers))
show_options(**settings)
```

## 3. Return values

`return` sends a value back to the caller and ends the function. A function with no `return` gives back `None`.

```python
def add(a, b):
    return a + b


def say_hi():
    print("hi")


result = add(2, 3)
print(result)

value = say_hi()
print(value)          # None
```

### Return several values

```python
def min_max_mean(numbers):
    return min(numbers), max(numbers), sum(numbers) / len(numbers)


result = min_max_mean([4, 8, 15])
print(result, type(result))        # a tuple

low, high, mean = min_max_mean([4, 8, 15])
print(low, high, round(mean, 2))
```

### Return early

```python
def safe_divide(a, b):
    if b == 0:
        return None          # early exit for a bad case
    return a / b


print(safe_divide(10, 4))
print(safe_divide(1, 0))
```

## 4. Docstrings and type hints

A **docstring** explains what a function does. **Type hints** document the expected types.
Python does not enforce hints, but editors and tools use them to catch mistakes.

```python
def estimate_cost(tokens: int, price_per_million: float = 3.0) -> float:
    """Return the cost in dollars for `tokens` at `price_per_million`.

    Args:
        tokens: Number of tokens used.
        price_per_million: Price for one million tokens (placeholder default).

    Returns:
        The cost, rounded to 4 decimal places.
    """
    return round(tokens * price_per_million / 1_000_000, 4)


print(estimate_cost(250_000))
help(estimate_cost)
```

## 5. Scope

```python
greeting = "global hello"      # global variable


def show():
    greeting = "local hello"   # a new, local variable
    print(greeting)


show()
print(greeting)                # unchanged
```

Variables created inside a function disappear when it ends. Pass values in and return results out instead of changing globals.

### The mutable default trap

A default value is created **once**, when the function is defined. A default list is shared by every call.

```python
def add_item_bad(item, items=[]):
    items.append(item)
    return items


print(add_item_bad("a"))
print(add_item_bad("b"))      # surprise: still contains "a"


def add_item_good(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items


print(add_item_good("a"))
print(add_item_good("b"))
```

## 6. Functions are values

```python
def shout(text):
    return text.upper()


def whisper(text):
    return text.lower()


def apply(func, text):
    return func(text)


print(apply(shout, "Hello"))
print(apply(whisper, "Hello"))
```

### Lambda functions

```python
square = lambda x: x ** 2
print(square(4))

people = [("Ada", 36), ("Linus", 28), ("Grace", 45)]
print(sorted(people, key=lambda person: person[1]))      # sort by age
print(list(map(lambda n: n * 2, [1, 2, 3])))
print(list(filter(lambda n: n > 1, [1, 2, 3])))
```

Use a lambda for a tiny, one-off expression. Use `def` for anything longer or reused.
