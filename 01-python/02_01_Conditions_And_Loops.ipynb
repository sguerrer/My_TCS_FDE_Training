# Conditions and Loops

Make decisions and repeat work. These two ideas control the flow of every program.

**You will learn to:**

- Branch with `if`, `elif` and `else`
- Combine tests with `and`, `or` and `not`
- Repeat work with `for` and `while`
- Control loops with `break`, `continue` and `else`

## 1. Conditions

```python
tokens_used = 7_500
limit = 10_000

if tokens_used > limit:
    print("Over the limit")
elif tokens_used > 0.8 * limit:
    print("Close to the limit")
else:
    print("Within budget")
```

Python uses **indentation** (4 spaces) to mark a block. A missing or uneven indent is a syntax error.

### Comparison and logical operators

| Operator | Meaning |
|---|---|
| `==` `!=` | Equal, not equal |
| `<` `<=` `>` `>=` | Order |
| `and` `or` `not` | Combine or flip tests |
| `in` `not in` | Membership |
| `is` `is not` | Same object (use for `None`) |

```python
age = 25
has_id = True

print(18 <= age < 65)                  # chained comparison
print(age >= 18 and has_id)
print(not has_id)
print("sonnet" in ["haiku", "sonnet"])

result = None
print(result is None)
```

### Truthiness in conditions

```python
name = ""

if not name:
    print("Name is empty")

items = [1, 2]
if items:
    print("List has items")
```

### One-line conditions and `match`

```python
score = 82
status = "pass" if score >= 50 else "fail"     # conditional expression
print(status)

command = "stop"
match command:                                  # Python 3.10+
    case "start":
        print("Starting")
    case "stop" | "quit":
        print("Stopping")
    case _:
        print("Unknown command")
```

## 2. `for` loops

```python
for model in ["haiku", "sonnet", "opus"]:
    print(model)

for i in range(3):             # 0, 1, 2
    print("round", i)

for i in range(2, 10, 3):      # start, stop, step
    print(i, end=" ")
print()
```

```python
names = ["Ada", "Linus", "Grace"]
scores = [90, 85, 99]

for index, name in enumerate(names, start=1):
    print(index, name)

for name, score in zip(names, scores):
    print(f"{name}: {score}")

for ch in "AI":
    print(ch)
```

### Accumulate a result

```python
total = 0
for price in [9.99, 4.5, 20.0]:
    total += price

print(f"Total: {total:.2f}")

# The same idea as a comprehension
prices = [9.99, 4.5, 20.0]
print([p for p in prices if p > 5])
```

## 3. `while` loops

Use `while` when you do not know in advance how many times to repeat.

```python
tokens_left = 1000
chunk = 300
round_number = 0

while tokens_left > 0:
    round_number += 1
    tokens_left -= chunk
    print(f"Round {round_number}: {max(tokens_left, 0)} tokens left")
```

> **Warning:** a `while` loop whose condition never becomes false runs forever. Make sure something inside the loop changes.

## 4. `break`, `continue` and `else`

```python
# break: stop the loop early
for n in range(10):
    if n == 4:
        break
    print("break loop:", n)

# continue: skip to the next item
for n in range(6):
    if n % 2 == 0:
        continue
    print("odd:", n)
```

```python
# for/else: the else block runs only if the loop did NOT break
for n in [3, 5, 7]:
    if n % 2 == 0:
        print("found an even number")
        break
else:
    print("no even numbers")
```

### Nested loops

```python
for row in range(1, 4):
    for col in range(1, 4):
        print(row * col, end="\t")
    print()
```

### Worked example: retry loop

Many AI applications call a service that sometimes fails, so they try again a limited number of times.
Here a function pretends to fail twice before it works.

```python
attempts_needed = 3
max_attempts = 5

for attempt in range(1, max_attempts + 1):
    print(f"Attempt {attempt}")
    if attempt >= attempts_needed:
        print("Success")
        break
else:
    print("Gave up")
```
