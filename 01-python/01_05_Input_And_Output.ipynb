## 1. Output with `print`

```python
print("Hello")
print("a", "b", "c")                 # items separated by a space
print("a", "b", "c", sep="-")        # custom separator
print("Loading", end="...")          # no new line
print("done")
```

```python
items = [("Notebook", 3, 2.5), ("Pen", 10, 0.99)]

print(f"{'Item':<10}{'Qty':>5}{'Price':>8}")
print("-" * 23)
for item, qty, price in items:
    print(f"{item:<10}{qty:>5}{price:>8.2f}")
```

## 2. Input with `input()`

`input()` **always returns a string**, even when the user types a number. Convert it yourself.

```python
name = input("What is your name? ")
print(f"Nice to meet you, {name}!")
```

```python
age_text = input("How old are you? ")
age = int(age_text)

print(f"Next year you will be {age + 1}.")
```

Typing something that is not a number would crash that cell. Check first, and ask again if needed.

```python
while True:
    text = input("Enter a whole number: ").strip()
    if text.isdigit():
        number = int(text)
        break
    print(f"'{text}' is not a whole number. Try again.")

print("Double:", number * 2)
```

## 3. Files

Open files with `with open(...)`. The `with` block closes the file for you, even if an error happens.

| Mode | Meaning |
|---|---|
| `"r"` | Read (default) |
| `"w"` | Write, replacing any existing content |
| `"a"` | Append to the end |

```python
with open("notes.txt", "w", encoding="utf-8") as file:
    file.write("First line\n")
    file.write("Second line\n")

with open("notes.txt", "a", encoding="utf-8") as file:
    file.write("Third line\n")

with open("notes.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```

```python
# Read line by line (good for large files)
with open("notes.txt", encoding="utf-8") as file:
    for number, line in enumerate(file, start=1):
        print(number, line.rstrip())

# Read all lines into a list
with open("notes.txt", encoding="utf-8") as file:
    lines = file.read().splitlines()
print(lines)
```

### `pathlib`: a modern way to handle paths

```python
from pathlib import Path

path = Path("notes.txt")

print(path.exists())
print(path.suffix, path.stem)
print(path.read_text(encoding="utf-8").count("line"))

path.write_text("Replaced\n", encoding="utf-8")
print(path.read_text(encoding="utf-8"))
```

```python
# Clean up the demo file
path.unlink()
print(path.exists())
```

## 4. A simple program

Put input, logic and output together: a small grade calculator.

```python
def letter_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    return "F"


print("Enter three test scores, separated by spaces.")
scores = [float(part) for part in input("Scores: ").split()]

average = sum(scores) / len(scores)
print(f"Scores : {scores}")
print(f"Average: {average:.1f}")
print(f"Grade  : {letter_grade(average)}")
```
