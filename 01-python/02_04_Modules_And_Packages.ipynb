## 1. Import code

A **module** is a single `.py` file. A **package** is a folder of modules. Python ships with a large **standard library**.

````python
import math
from random import randint, choice
from datetime import datetime, timedelta
import os as operating_system          # alias

print(math.sqrt(81))
print(randint(1, 6) in range(1, 7))
print(choice(["a", "b", "c"]) in "abc")
print((datetime(2026, 10, 26) + timedelta(days=7)).date())
print(operating_system.name)
````

Prefer explicit imports. Avoid `from module import *`: it hides where names come from.

### A short tour of the standard library

| Module | Use for |
|---|---|
| `math`, `statistics` | Maths and simple statistics |
| `random` | Random numbers and choices |
| `datetime` | Dates and times |
| `pathlib`, `os` | Paths and the operating system |
| `json`, `csv` | Reading and writing data files |
| `collections` | `Counter`, `defaultdict`, `deque` |
| `re` | Regular expressions |
| `logging` | Application logs |

````python
from collections import Counter, defaultdict
import statistics

print(Counter("mississippi").most_common(2))

groups = defaultdict(list)
for word in ["apple", "avocado", "banana", "blueberry"]:
    groups[word[0]].append(word)
print(dict(groups))

print(statistics.mean([2, 4, 9]), statistics.median([2, 4, 9]))
````

## 2. Create your own module

The next cell writes a small file called `ai_helpers.py` next to this notebook, then imports it.

````python
from pathlib import Path
import importlib

Path("ai_helpers.py").write_text(
    """\"\"\"Small helpers for AI scripts.\"\"\"

TOKENS_PER_WORD = 1.3


def estimate_tokens(text: str) -> int:
    \"\"\"Return a rough token estimate for `text`.\"\"\"
    return round(len(text.split()) * TOKENS_PER_WORD)


if __name__ == "__main__":
    print("Running as a script:", estimate_tokens("hello world"))
""",
    encoding="utf-8",
)

importlib.invalidate_caches()           # make Python notice the new file
````

````python
import ai_helpers

print(ai_helpers.estimate_tokens("Tokens are the units a model reads"))
print(ai_helpers.TOKENS_PER_WORD)

from ai_helpers import estimate_tokens
print(estimate_tokens("short"))
````

### `if __name__ == "__main__"`

Python sets `__name__` to `"__main__"` only when a file is **run** directly, and to the module name when it is **imported**.
Code under this guard runs as a script but stays quiet on import.

````python
print(ai_helpers.__name__)
print(__name__)                     # the notebook itself is "__main__"
````

````python
# Run the file as a script to trigger the guarded block
import subprocess, sys

completed = subprocess.run([sys.executable, "ai_helpers.py"], capture_output=True, text=True)
print(completed.stdout.strip())
````

## 3. Create a package

A package is a folder with an `__init__.py` file. It groups related modules.

```text
shop_tools/
├── __init__.py
├── prices.py
└── text.py
```

````python
pkg = Path("shop_tools")
pkg.mkdir(exist_ok=True)

(pkg / "__init__.py").write_text(
    "from .prices import with_tax\n",
    encoding="utf-8",
)
(pkg / "prices.py").write_text(
    "def with_tax(amount, rate=0.2):\n"
    "    return round(amount * (1 + rate), 2)\n",
    encoding="utf-8",
)
(pkg / "text.py").write_text(
    "def slug(title):\n"
    "    return title.strip().lower().replace(' ', '-')\n",
    encoding="utf-8",
)

importlib.invalidate_caches()
````

````python
import shop_tools
from shop_tools import text
from shop_tools.prices import with_tax

print(shop_tools.with_tax(100))        # exposed by __init__.py
print(with_tax(50, rate=0.1))
print(text.slug("  Claude Architect Program "))
````

````python
# Clean up the files created by this notebook
import shutil

shutil.rmtree("shop_tools")
Path("ai_helpers.py").unlink()
shutil.rmtree("__pycache__", ignore_errors=True)
print("cleaned up")
````

## 4. External packages with `pip`

Packages written by others live on **PyPI**. Install them with `pip` inside your virtual environment.
Run these in a **terminal** with `.venv` active:

```bash
pip install requests            # install
pip install "requests>=2.31"    # with a version rule
pip list                        # see what is installed
pip show requests               # details about one package
pip freeze > requirements.txt   # save exact versions
pip install -r requirements.txt # recreate on another machine
pip uninstall requests          # remove
```

In a notebook, `%pip install <package>` installs into the active kernel's environment.

Check from Python which environment is running and which version of a package is installed.

````python
import sys
from importlib.metadata import version, PackageNotFoundError

print("Python    :", sys.version.split()[0])
print("Executable:", sys.executable)
print("In a venv :", sys.prefix != sys.base_prefix)

for package in ["pip", "anthropic", "not-a-real-package"]:
    try:
        print(f"{package:<20}", version(package))
    except PackageNotFoundError:
        print(f"{package:<20}", "not installed")
````

> **Tip:** if `In a venv` is `False`, the notebook is not using your `.venv` kernel. Select it from the kernel picker.
