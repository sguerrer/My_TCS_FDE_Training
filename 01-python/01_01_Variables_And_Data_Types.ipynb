## 1. Variables

A **variable** is a name that points to a value. Create one with `=`.
Python works out the type for you, so you never declare it.

```python
name = "Ada"
age = 30
height_m = 1.68
is_student = False

print(name, age, height_m, is_student)
```

A variable can be re-assigned, even to a different type.

```python
score = 10
print(score, type(score))

score = "ten"
print(score, type(score))
```

Assign several variables on one line, or give them the same value.

```python
x, y, z = 1, 2, 3
a = b = c = 0

print(x, y, z)
print(a, b, c)

# Swap two variables without a temporary
x, y = y, x
print(x, y)
```

### Naming rules

| Rule | Valid | Invalid |
|---|---|---|
| Start with a letter or `_` | `total`, `_cache` | `2nd_place` |
| Use letters, digits and `_` only | `max_tokens` | `max-tokens` |
| Avoid reserved words | `class_name` | `class`, `for`, `if` |
| Names are case-sensitive | `Age` and `age` differ | |

**Convention:** use `snake_case` for variables and `UPPER_SNAKE_CASE` for constants.

```python
MAX_TOKENS = 1024        # constant: by convention, never re-assigned
user_name = "Ada"        # snake_case variable

print(MAX_TOKENS, user_name)
```

## 2. Core data types

```python
values = [42, 3.14, True, "hello", None]

for value in values:
    print(f"{value!r:>8}  ->  {type(value).__name__}")
```

| Type | Example | Use for |
|---|---|---|
| `int` | `42` | Whole numbers (counts, indexes) |
| `float` | `3.14` | Decimal numbers (prices, scores) |
| `bool` | `True` / `False` | Yes/no decisions |
| `str` | `"hello"` | Text |
| `NoneType` | `None` | "No value yet" |

### Booleans and truthiness

Any value can be used in a condition. These values are **falsy**: `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}`.
Everything else is **truthy**.

```python
for value in [0, 1, "", "text", [], [0], None]:
    print(f"{value!r:>7} is {bool(value)}")
```

## 3. Type conversion

Use `int()`, `float()`, `str()` and `bool()` to convert between types.

```python
print(int("42") + 8)        # str -> int
print(float("3.5") * 2)     # str -> float
print(str(100) + " tokens") # int -> str
print(int(7.9))             # float -> int drops the decimal (does not round)
print(round(7.9))           # round() rounds to the nearest whole number
```

A conversion that does not make sense raises an error. You will handle errors properly on Day 2.

```python
try:
    int("hello")
except ValueError as error:
    print("Cannot convert:", error)
```

## 4. Check a type

```python
max_tokens = 1024

print(type(max_tokens))
print(isinstance(max_tokens, int))
print(isinstance(max_tokens, (int, float)))   # matches either type
```
