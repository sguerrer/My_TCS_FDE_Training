# Dictionaries and Sets

Store data by name with dictionaries, and unique values with sets.

**You will learn to:**

- Create, read, update and loop over dictionaries
- Nest dictionaries and lists to model real data
- Use sets to remove duplicates and compare groups
- Choose the right collection for a task

## 1. Dictionaries

A **dictionary** maps **keys** to **values**. Use it when you look things up by name rather than position.

```python
model = {
    "name": "sonnet",
    "max_tokens": 1024,
    "temperature": 0.7,
}

print(model["name"])
print(len(model))
print("temperature" in model)     # checks keys
```

### Read safely with `get`

```python
print(model.get("max_tokens"))
print(model.get("top_p"))              # missing key -> None
print(model.get("top_p", 1.0))         # missing key -> your default

try:
    model["top_p"]
except KeyError as error:
    print("KeyError:", error)
```

### Add, update and delete

```python
model["top_p"] = 0.9                   # add
model["temperature"] = 0.2             # update
model.update({"stream": True, "max_tokens": 2048})
print(model)

removed = model.pop("stream")          # delete and return
del model["top_p"]                     # delete
print(removed, model)
```

### Loop over a dictionary

```python
for key in model:
    print("key:", key)

for value in model.values():
    print("value:", value)

for key, value in model.items():
    print(f"{key} = {value}")
```

### Nested data

Real data, such as a chat history for an LLM, is usually a **list of dictionaries**.
You will use this exact shape with the Claude API later in the course.

```python
messages = [
    {"role": "user", "content": "What is a token?"},
    {"role": "assistant", "content": "A small chunk of text the model reads."},
    {"role": "user", "content": "Give me an example."},
]

print(messages[0]["content"])
print(len(messages), "messages")

user_messages = [m["content"] for m in messages if m["role"] == "user"]
print(user_messages)
```

```python
config = {
    "model": {"name": "sonnet", "max_tokens": 1024},
    "tags": ["demo", "day-1"],
}

print(config["model"]["name"])
print(config["tags"][1])
```

### Dictionary comprehensions and merging

```python
words = ["claude", "python", "ai"]
lengths = {word: len(word) for word in words}
print(lengths)

defaults = {"temperature": 1.0, "max_tokens": 512}
overrides = {"temperature": 0.2}
print(defaults | overrides)    # right side wins on clashes (Python 3.9+)
```

### Count with a dictionary

```python
text = "the cat and the hat and the bat"
counts = {}

for word in text.split():
    counts[word] = counts.get(word, 0) + 1

print(counts)
print(sorted(counts.items(), key=lambda pair: pair[1], reverse=True))
```

## 2. Sets

A **set** is an unordered collection of **unique** values. Create one with `{}` and values, or with `set()`.

```python
tags = {"ai", "python", "ai", "cloud"}
print(tags)               # duplicates are gone; order is not guaranteed

empty_set = set()         # {} would create an empty dict
print(type(empty_set), type({}))
```

```python
tags.add("aws")
tags.discard("cloud")     # no error if missing (remove() raises KeyError)
print(tags)
print("python" in tags)   # fast membership test
```

### Remove duplicates from a list

```python
emails = ["a@x.com", "b@x.com", "a@x.com", "c@x.com", "b@x.com"]

unique = set(emails)
print(len(emails), "->", len(unique))
print(sorted(unique))
```

### Compare two groups

```python
day1 = {"alice", "bob", "carol"}
day2 = {"bob", "carol", "dave"}

print(day1 | day2)   # union: in either
print(day1 & day2)   # intersection: in both
print(day1 - day2)   # difference: only in day1
print(day1 ^ day2)   # symmetric difference: in exactly one
print({"bob"} <= day1)   # subset test
```

## Choose a collection

| Need | Use |
|---|---|
| Ordered items you will change | `list` |
| Fixed group of values | `tuple` |
| Look up values by a name | `dict` |
| Unique values or set maths | `set` |
