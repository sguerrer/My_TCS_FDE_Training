## 1. Lists

A **list** is an ordered, **mutable** collection. Create one with square brackets.

```python
models = ["haiku", "sonnet", "opus"]
mixed = [1, "two", 3.0, True]

print(models)
print(len(models))
print(models[0], models[-1])
print(models[0:2])
```

### Change a list

```python
tasks = ["write", "test"]

tasks.append("deploy")          # add to the end
tasks.insert(0, "plan")         # add at a position
tasks.extend(["review", "ship"])# add several
print(tasks)

tasks.remove("test")            # remove by value
last = tasks.pop()              # remove and return the last item
print(last, tasks)

tasks[0] = "design"             # replace by index
print(tasks)
```

### Search and sort

```python
scores = [72, 95, 88, 61, 95]

print("deploy" in ["plan", "deploy"])
print(scores.index(88))         # position of the first match
print(scores.count(95))

print(sorted(scores))                  # returns a new sorted list
print(sorted(scores, reverse=True))
scores.sort()                          # sorts in place and returns None
print(scores)
print(max(scores), min(scores), sum(scores) / len(scores))
```

### Aliases and copies

`b = a` does **not** copy a list. Both names point to the same list.

```python
a = [1, 2, 3]
b = a              # alias
c = a.copy()       # real copy

a.append(4)
print("a:", a)
print("b:", b)     # changed too
print("c:", c)     # unchanged
```

### Loop with an index

```python
steps = ["load", "clean", "train"]

for number, step in enumerate(steps, start=1):
    print(f"{number}. {step}")
```

### List comprehensions

A compact way to build a new list: `[expression for item in items if condition]`.

```python
numbers = [1, 2, 3, 4, 5, 6]

squares = [n ** 2 for n in numbers]
evens = [n for n in numbers if n % 2 == 0]
labels = [f"item-{n}" for n in numbers[:3]]

print(squares)
print(evens)
print(labels)
```

### Nested lists

```python
grid = [
    [1, 2, 3],
    [4, 5, 6],
]

print(grid[1][2])                          # row 1, column 2
print([row[0] for row in grid])            # first column
```

## 2. Tuples

A **tuple** is an ordered, **immutable** collection. Use it for a fixed group of values that belong together.

```python
point = (10, 20)
person = ("Ada", 30, "London")

print(point[0], person[1])
print(len(person))

try:
    point[0] = 99
except TypeError as error:
    print("Error:", error)
```

A tuple with one item needs a trailing comma.

```python
print(type((5,)))   # tuple
print(type((5)))    # just an int in brackets
```

### Unpacking

```python
name, age, city = ("Ada", 30, "London")
print(name, age, city)

first, *rest = [1, 2, 3, 4]
print(first, rest)

# Functions often return tuples
low, high = min(5, 2, 9), max(5, 2, 9)
print(low, high)
```

### List or tuple?

| | List | Tuple |
|---|---|---|
| Syntax | `[1, 2]` | `(1, 2)` |
| Changeable | Yes | No |
| Typical use | A collection that grows or changes | A fixed record, such as `(x, y)` |
| Can be a dict key | No | Yes (if its items are immutable) |
