## 1. Strings

```python
prompt = "Summarise this document"

print(prompt)
print(len(prompt))            # number of characters
print(prompt.upper())
print(prompt.lower())
print(prompt.title())
```

### Indexing and slicing

Positions start at `0`. Negative positions count from the end.
A slice `s[start:stop]` includes `start` and excludes `stop`.

```python
word = "Claude"

print(word[0])      # first character
print(word[-1])     # last character
print(word[0:3])    # 'Cla'
print(word[2:])     # 'aude'
print(word[::-1])   # reversed
```

Strings are **immutable**: you cannot change a character in place. Build a new string instead.

```python
word = "Claude"

try:
    word[0] = "K"
except TypeError as error:
    print("Error:", error)

new_word = "K" + word[1:]
print(new_word)
```

### Useful string methods

```python
raw = "   Hello, World!   "

print(repr(raw.strip()))                 # remove surrounding whitespace
print(raw.strip().replace("World", "Python"))
print("a,b,c".split(","))                # str -> list
print("-".join(["2026", "10", "26"]))    # list -> str
print("hello".startswith("he"), "hello".endswith("lo"))
print("hello".find("l"), "hello".count("l"))
print("Python" in "I love Python")
```

### f-strings

Put `f` before the quotes and wrap expressions in `{}`.
Add a format after `:` to control width, decimals and separators.

```python
name = "Ada"
price = 1234.5
ratio = 0.8765

print(f"Hello, {name}!")
print(f"Price: {price:,.2f}")       # thousands separator, 2 decimals
print(f"Ratio: {ratio:.1%}")        # percentage
print(f"{name:>10}|{name:<10}|{name:^10}|")   # right, left, centre
print(f"2 + 3 = {2 + 3}")
```

### Multi-line strings and escapes

```python
system_prompt = """You are a helpful assistant.
Answer in one short paragraph."""

print(system_prompt)
print("Column A\tColumn B")      # \t is a tab
print("Line 1\nLine 2")          # \n is a new line
print(r"C:\new_folder\test")     # r"..." keeps backslashes literal
```

## 2. Numbers

```python
a, b = 17, 5

print(a + b)    # addition
print(a - b)    # subtraction
print(a * b)    # multiplication
print(a / b)    # true division -> always a float
print(a // b)   # floor division
print(a % b)    # remainder
print(a ** 2)   # power
print(divmod(a, b))
```

Operators follow the usual order: `**`, then `* / // %`, then `+ -`. Use brackets to be explicit.

```python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(-3 ** 2, (-3) ** 2)
```

```python
import math

print(abs(-7.5))
print(round(3.14159, 2))
print(max(3, 9, 4), min(3, 9, 4))
print(sum([1, 2, 3, 4]))
print(math.sqrt(144))
print(math.ceil(2.1), math.floor(2.9))
print(math.pi)
```

### Float precision

Computers store decimals in binary, so some values are only approximate.
Compare floats with `math.isclose`, and round before you display them.

```python
import math

print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
print(round(0.1 + 0.2, 2))
```

## Worked example: estimate an API cost

Pricing for language models is quoted per million tokens. Combine numbers and f-strings to estimate a bill.
The prices below are **made-up placeholders**; look up current prices before you budget.

```python
input_tokens = 12_000            # underscores make big numbers readable
output_tokens = 3_500
price_in_per_million = 3.00      # placeholder, USD
price_out_per_million = 15.00    # placeholder, USD

cost = (input_tokens * price_in_per_million + output_tokens * price_out_per_million) / 1_000_000

print(f"Input tokens : {input_tokens:>8,}")
print(f"Output tokens: {output_tokens:>8,}")
print(f"Estimated cost: ${cost:.4f}")
```
